#### Euler's Method

Below you can find code to run Euler's Method to approximate the solution of
$$ \frac{dx}{dt} = f(t,x) $$
with the initial condition $x(t_0) = x_0$.

The method will approximate the solution for the points $t_i = t_0 + i h$ for $i = 0, \ldots, n$ where the following information needs to be defined:
- `t_span`$ = ( t_0, t_n)$
- `h` where $n = \frac{t_n - t_0}{h}$ 
- `x0` = $x_0$
- `f(t,x)`

After running Euler's Method, the approximation will be stored as:
- `t` a list containing all the points $t_i$
- `x` a list containing the approximation of $x(t_i)$

The code will also plot the solution found.

```python
import numpy as np
import matplotlib.pyplot as plt

def euler_method(f, t_span, x0, h):
    """
    Approximates the solution of a first-order ODE using Euler's method.

    Args:
        f: The function defining the ODE, x' = f(t, x).
        t_span: A tuple (t0, tf) representing the start and end times.
        x0: The initial condition, x(t0).
        h: The step size.

    Returns:
        A tuple containing arrays of t and x values.
    """

    t0, tf = t_span
    t = np.arange(t0, tf + h, h)  # Create t values with correct endpoint
    n = len(t)
    x = np.zeros(n)
    x[0] = x0

    for i in range(n - 1):
        x[i+1] = x[i] + h * f(t[i], x[i])

    return t, x

# Example usage: Solve dx/dt = - k(t) x + t, x(0) = 1, from t=0 to t=100


# We allow k to change with t by defining a function k(t)
def k(t):
    return 3


def f(t, x):
    return - k(t) * x + t

    
t_span = (0, 100)
x0 = 1
h = 1  # Step size


t, x = euler_method(f, t_span, x0, h)


"""Plots the solution obtained from Euler's method."""
plt.plot(t, x, linestyle='-', label='Euler Approximation')
plt.xlabel('tau')
plt.ylabel('x')
plt.grid(True)
plt.legend()
plt.show()


print('Values of t')
print(t)
print('\n')
print('Values of x')
print(x)
```
